## STEP 2 — Write playground_config

```python
%%writefile ~/playground_config.py
import os
from langchain_ollama import ChatOllama, OllamaEmbeddings

OLLAMA_BASE_URL = "http://ollama.ollama:11434"
OLLAMA_MODEL = "llama3.2"
OLLAMA_EMBED_MODEL = "nomic-embed-text"

def get_llm(**kwargs):
    return ChatOllama(model=OLLAMA_MODEL, base_url=OLLAMA_BASE_URL, **kwargs)

def get_embeddings(**kwargs):
    return OllamaEmbeddings(model=OLLAMA_EMBED_MODEL, base_url=OLLAMA_BASE_URL, **kwargs)

def status():
    print(f"Backend : ollama")
    print(f"URL     : {OLLAMA_BASE_URL}")
    print(f"Model   : {OLLAMA_MODEL}")
```

```python
import sys, os
sys.path.insert(0, os.path.expanduser("~"))
from playground_config import get_llm, status
status()
```

## STEP 3 — Run a simple LangChain chain

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = get_llm()

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant. Answer in one sentence."),
    ("human", "{question}")
])

chain = prompt | llm | StrOutputParser()

response = chain.invoke({"question": "What is LangChain?"})
print(response)
```

## STEP 4 — Stream a response

```python
for chunk in chain.stream({"question": "What is RAG in AI?"}): 
    print(chunk, end="", flush=True)
```
